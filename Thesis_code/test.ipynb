```python
import benchmarks_CEC as bc
import jax.numpy as jnp

params = bc.get_cec_params(dim=30)

simple_ids = [
    bc.CEC_BENT_CIGAR,
    bc.CEC_ZAKHAROV,
    bc.CEC_RASTRIGIN,
    bc.CEC_SCAFFERS_F6,
]

for obj_id in simple_ids:
    p = params[obj_id]
    x_star = p["shift"]
    val = bc.evaluate_objective(x_star, obj_id)
    print(obj_id, bc.get_objective_name(obj_id), "value at shift =", float(val), "expected bias =", float(p["bias"]))
```
